```python
import numpy as np
from sklearn import preprocessing
```

```python
feature = np.array( [[-500.0],
[-100.1],
[0],
[100.1],
[900.9]])
print(feature)
```

```python
minmax_scale = preprocessing.MinMaxScaler(feature_range=(0, 1)) 

# In order to bring the column of numbers to a narrow range, it is going to take the minimum value in the column and maximum value in the column. Using those, it is going to bring the numbers to a narrow range
# (each value - minimum value) / (maximum value - minimum value)
```

```python
scaled_feature = minmax_scale.fit_transform(feature)
scaled_feature
```
