```python
# Import the pandas library for data manipulation
import pandas as pd
```

```python
# Create an empty DataFrame
df = pd.DataFrame()
```

```python
# Add columns to the DataFrame with sample data
df["Name"] = ["Ahmed", "Ali"]
df["Age"] = [16, 17]
df["Driver"] = [False, False]
df
```

```python
# Create a new row as a Series with matching column names
new_person = pd.Series(["Mohamed", 18, True], index=df.columns)
df
```

```python
# Append the new row to the DataFrame
df = pd.concat([df, new_person.to_frame().T], ignore_index=True)
df
```

```python
# Alternatively, you can also use the .loc method to append the new row
df.loc[len(df)] = new_person
df
```

```python
# replace all the "female" values in the "Sex" column with "women" word
titanic["Sex"] = titanic["Sex"].replace("female","women")
titanic
```

```python
titanic["Sex"] = titanic["Sex"].replace("male","M")
titanic
```

```python
titanic
```

```python
# replace all the "M" values in the "Sex" column with "male" word and all the "women" values with "female" word
titanic["Sex"] = titanic["Sex"].replace(["M","women"],["male","female"])
titanic
```

```python
titanic["Sex"] = titanic["Sex"].replace(["male","female"],["M","F"])
titanic
```

```python
titanic["Survived"] = titanic["Survived"].replace({0: "Died", 1: "Survived"})
titanic
```

```python
titanic = titanic.replace(1, "one")
titanic
```

```python
titanic.rename(columns={"Pclass": "Passenger Class", "Name": "Full Name"}, inplace=True)
titanic
#titanic = titanic.rename(columns={"Pclass": "Passenger Class"}) same thing as above
```

```python
titanic.info()
```

```python
# Convert the "Age" column to numeric values, coercing any non-numeric values to NaN
titanic['Age'] = pd.to_numeric(titanic['Age'], errors='coerce')

print(f"max age: {titanic['Age'].max()}")
print(f"min age: {titanic['Age'].min()}")
print(f"mean age: {titanic['Age'].mean()}")
print(f"sum age: {titanic['Age'].sum()}")
print(f"count age: {titanic['Age'].count()}")
print(f"median age: {titanic['Age'].median()}")
print(f"std age: {titanic['Age'].std()}")

```
