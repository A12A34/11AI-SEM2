```python
# Complete code for Data Wrangling Pt 1
import pandas as pd
import numpy as np
from pathlib import Path

csv_path = Path("Titanic-Dataset.csv")
if not csv_path.exists():
    csv_path = Path.cwd() / "school" / "ML" / "Final revision" / "Titanic-Dataset.csv"

# Creating a DataFrame and adding rows
people_df = pd.DataFrame()
people_df["Name"] = ["Ahmed", "Ali"]
people_df["Age"] = [16, 17]
people_df["Driver"] = [False, False]

new_person = pd.Series(["Mohammed", 18, True], index=people_df.columns)
people_df = pd.concat([people_df, new_person.to_frame().T], ignore_index=True)
people_df.loc[len(people_df)] = ["Fatima", 19, True]
print("People DataFrame:")
display(people_df)

# Reading and exploring Titanic data
titanic = pd.read_csv(csv_path)
print("Titanic preview:")
display(titanic.head())
print("Info:")
titanic.info()
print("Describe:")
display(titanic.describe(include="all"))
print("Shape:", titanic.shape)

# Filtering rows
print("Top two male passengers:")
display(titanic[titanic["Sex"] == "male"].head(2))
print("Female passengers aged 60 or older:")
display(titanic[(titanic["Sex"] == "female") & (titanic["Age"] >= 60)])

# Replacing values without damaging the original data
clean_titanic = titanic.copy(deep=True)
clean_titanic["Gender"] = clean_titanic["Sex"].replace({"female": "Woman", "male": "Man"})
clean_titanic["DeadOrAlive"] = clean_titanic["Survived"].replace({0: "dead", 1: "alive"})
clean_titanic.rename(columns={"Pclass": "Passenger Class"}, inplace=True)
print("After replace and rename:")
display(clean_titanic.head())

# Statistics and counts
clean_titanic["Age"] = pd.to_numeric(clean_titanic["Age"], errors="coerce")
print("Maximum Age:", clean_titanic["Age"].max())
print("Minimum Age:", clean_titanic["Age"].min())
print("Mean Age:", clean_titanic["Age"].mean())
print("Unique genders:", clean_titanic["Gender"].unique())
print("Gender counts:")
display(clean_titanic["Gender"].value_counts())
print("Missing ages:", clean_titanic["Age"].isna().sum())
display(clean_titanic[clean_titanic["Age"].isna()].head())

# Reading with custom missing values
titanic_with_na = pd.read_csv(csv_path, na_values=[np.nan, "NONE", -999])
display(titanic_with_na.head())

# Deleting columns, rows, and duplicates on copies
column_drop_df = clean_titanic.copy(deep=True)
column_drop_df.drop(columns=["Age"], inplace=True)
display(column_drop_df.head(10))

multi_drop_df = clean_titanic.copy(deep=True)
multi_drop_df.drop(columns=["SibSp", "Parch"], inplace=True)
display(multi_drop_df.head(10))

first_column_drop_df = clean_titanic.copy(deep=True)
first_column_drop_df.drop(columns=[first_column_drop_df.columns[0]], inplace=True)
display(first_column_drop_df.head(10))

row_drop_df = clean_titanic.copy(deep=True)
row_drop_df.drop(index=[0, 1, 2], inplace=True)
display(row_drop_df.head(10))

duplicate_drop_df = clean_titanic.copy(deep=True)
duplicate_drop_df.drop_duplicates(inplace=True)
print("Original rows:", len(clean_titanic))
print("Rows after duplicate removal:", len(duplicate_drop_df))
display(duplicate_drop_df.head(10))

gender_unique_first = clean_titanic.copy(deep=True)
gender_unique_first.drop_duplicates(subset=["Gender"], keep="first", inplace=True)
display(gender_unique_first.head(10))

gender_unique_last = clean_titanic.copy(deep=True)
gender_unique_last.drop_duplicates(subset=["Gender"], keep="last", inplace=True)
display(gender_unique_last.head(10))
```
