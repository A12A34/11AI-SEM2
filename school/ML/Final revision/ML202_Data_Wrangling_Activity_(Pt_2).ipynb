```python
# Complete code for Data Wrangling Activity Pt 2
from sklearn import datasets
import pandas as pd

# Load Iris dataset and create DataFrame
iris = datasets.load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["target"] = iris.target

print("Dataset shape:", iris.data.shape)
display(df.head())

# Explore data
print("Info:")
df.info()
print("Describe:")
display(df.describe())
print("Any missing values:", df.isnull().values.any())
print("Missing values by column:")
display(df.isnull().sum())

# Filter rows
print("Sepal length >= 4.3:")
display(df[df["sepal length (cm)"] >= 4.3].head())
print("Sepal length > 5 and petal length > 4.35:")
display(df[(df["sepal length (cm)"] > 5) & (df["petal length (cm)"] > 4.35)].head())

# Rename columns
renamed_df = df.rename(columns={
    "sepal length (cm)": "sepal_length",
    "sepal width (cm)": "sepal_width",
    "petal length (cm)": "petal_length",
    "petal width (cm)": "petal_width",
})
display(renamed_df.head())

# Statistics and counts
print("Maximum sepal length:", renamed_df["sepal_length"].max())
print("Minimum sepal width:", renamed_df["sepal_width"].min())
print("Petal length count:", renamed_df["petal_length"].count())
print("Petal length length:", len(renamed_df["petal_length"]))
print("Unique petal lengths:")
display(pd.Series(renamed_df["petal_length"].unique()))
print("Petal length value counts:")
display(renamed_df["petal_length"].value_counts().head())

# Drop column, rows, and duplicates on copies
without_petal_length = renamed_df.copy(deep=True)
without_petal_length.drop("petal_length", axis=1, inplace=True)
display(without_petal_length.head())

without_first_rows = renamed_df.copy(deep=True)
without_first_rows.drop([0, 1], axis=0, inplace=True)
display(without_first_rows.head())

no_duplicates = renamed_df.copy(deep=True)
no_duplicates.drop_duplicates(inplace=True)
print("Rows before duplicates:", len(renamed_df))
print("Rows after duplicates:", len(no_duplicates))

# Deep copy
safe_copy = renamed_df.copy(deep=True)
display(safe_copy.head())
```
