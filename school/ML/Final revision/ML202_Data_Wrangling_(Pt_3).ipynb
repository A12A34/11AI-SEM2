```python
# Grouping rows by values

import pandas as pd
```

```python
dataframe = pd.read_csv("Titanic-Dataset.csv")
dataframe.head(10)
```

```python
dataframe.groupby('Sex').mean(numeric_only=True)
```

```python
# Grouping rows by time

import numpy as np
```

```python
time_index = pd.date_range('06/06/2017', periods=100000, freq='30S')
time_index
```

```python
# Creating a dataframe

df1 = pd.DataFrame(index = time_index)
df1
```

```python
# Creating a column of random values

df1['Sale_Amount'] = np.random.randint(1, 10, 100000) # Random integers between 1 and 10
df1
```

```python
# Grouping rows by week, calculating sum per week

df1.resample('W').sum()
```

```python
# Grouping by two weeks, calculating mean

df1.resample('2W').mean()
```

```python
# Grouping by month, counting rows

df1.resample('M').count()
```

```python
# Looping over a column

for name in dataframe['Name'][0:2]:
  print(name.upper())
```

```python
# A different way to loop over a column
# List comprehension

[name.upper() for name in dataframe['Name'][0:2]]
```

```python
# Applying a function over all elements in a column

def uppercase(x):
  return x.upper()
```

```python
# Applying the function created above to the column 'Name'

dataframe['Name'].apply(uppercase)[0:2]
```

```python
# Applying function to groups

dataframe.groupby('Sex').apply(lambda x: x.count())
```

```python
# Concatenating dataframes

df2 = {"id": ["1", "2", "3"],
       "first_name": ["Ahmed", "Ali", "Hasan"],
       "last_name": ["Hasan", "Ahmed", "Ali"]}
df2 = pd.DataFrame(df2, columns = ["id", "first_name", "last_name"])

df2
```

```python
# Creating a second dataframe

df3 = {"id": ["4", "5", "6"],
       "first_name": ["Ebrahim", "Isa", "Mohammed"],
       "last_name": ["Mohammed", "Ebrahim", "Isa"]}

df3 = pd.DataFrame(df3, columns = ["id", "first_name", "last_name"])

df3
```

```python
# Concatenating Dataframes by rows

df2 = pd.concat([df2, df3], axis = 0)
df2
```

```python
# Creating a row

row = pd.Series([10, "Hussain", "Ali"], index = ["id", "first_name", "last_name"])
```

```python
# Appending row

pd.concat([df2, row.to_frame().T], ignore_index = True)
```

```python
# Merging dataframes

# Creating dataframe

employee_data = {"id": ["1", "2", "3", "4"],
       "first_name": ["Ebrahim", "Isa", "Mohammed", "Ahmed"],
       "last_name": ["Mohammed", "Ebrahim", "Isa", "Hussan"]}

employee_data = pd.DataFrame(employee_data, columns = ["id", "first_name", "last_name"])
```

```python
employee_data
```

```python
# Creating second dataframe

sales_data = {"employee_id": ["3", "4", "5", "6"],
              "total_sales": [23456, 2512, 2345, 1455]}

sales_df = pd.DataFrame(sales_data, columns = ["employee_id", "total_sales"])
```

```python
sales_df
```

```python
# Merging dataframes

pd.merge(employee_data, sales_df, left_on = "id", right_on = "employee_id")
```

```python
pd.merge(employee_data, sales_df, left_on = "id", right_on = "employee_id", how = "outer")
```

```python
pd.merge(employee_data, sales_df, left_on="id", right_on="employee_id", how="inner")
```

```python
# Complete code for Data Wrangling Pt 3
import pandas as pd
import numpy as np
from pathlib import Path

csv_path = Path("Titanic-Dataset.csv")
if not csv_path.exists():
    csv_path = Path.cwd() / "school" / "ML" / "Final revision" / "Titanic-Dataset.csv"

# Grouping rows by values
titanic = pd.read_csv(csv_path)
display(titanic.head(10))
print("Mean numeric values by sex:")
display(titanic.groupby("Sex").mean(numeric_only=True))
print("Mean numeric values by survived:")
display(titanic.groupby("Survived").mean(numeric_only=True))

# Grouping rows by time
time_index = pd.date_range("2017-06-06", periods=100000, freq="30s")
sales_time_df = pd.DataFrame(index=time_index)
sales_time_df["Sale_Amount"] = np.random.randint(1, 10, 100000)
display(sales_time_df.head())
print("Weekly sales sum:")
display(sales_time_df.resample("W").sum().head())
print("Two-week sales mean:")
display(sales_time_df.resample("2W").mean().head())
print("Monthly row count:")
display(sales_time_df.resample("ME").count().head())

# Looping over a column
print("Uppercase names with loop:")
for name in titanic["Name"].head(2):
    print(name.upper())

uppercase_names = [name.upper() for name in titanic["Name"].head(2)]
print("Uppercase names with list comprehension:", uppercase_names)

def uppercase(value):
    return value.upper()

print("Uppercase names with apply:")
display(titanic["Name"].apply(uppercase).head(2))
print("Count by sex:")
display(titanic.groupby("Sex").count())

# Concatenating DataFrames
students_1 = pd.DataFrame({
    "id": ["1", "2", "3"],
    "first_name": ["Ahmed", "Ali", "Hasan"],
    "last_name": ["Hasan", "Ahmed", "Ali"],
})
students_2 = pd.DataFrame({
    "id": ["4", "5", "6"],
    "first_name": ["Ebrahim", "Isa", "Mohammed"],
    "last_name": ["Mohammed", "Ebrahim", "Isa"],
})
all_students = pd.concat([students_1, students_2], axis=0, ignore_index=True)
display(all_students)

row = pd.Series(["10", "Hussain", "Ali"], index=["id", "first_name", "last_name"])
all_students_with_row = pd.concat([all_students, row.to_frame().T], ignore_index=True)
display(all_students_with_row)

# Merging DataFrames
employee_df = pd.DataFrame({
    "id": ["1", "2", "3", "4"],
    "first_name": ["Ebrahim", "Isa", "Mohammed", "Ahmed"],
    "last_name": ["Ali", "Ahmed", "Hasan", "Mohammed"],
})
sales_df = pd.DataFrame({
    "employee_id": ["3", "4", "5", "6"],
    "total_sales": [23456, 2512, 2345, 1455],
})

display(employee_df)
display(sales_df)
print("Default inner merge:")
display(pd.merge(employee_df, sales_df, left_on="id", right_on="employee_id"))
print("Outer merge:")
display(pd.merge(employee_df, sales_df, left_on="id", right_on="employee_id", how="outer"))
print("Inner merge:")
display(pd.merge(employee_df, sales_df, left_on="id", right_on="employee_id", how="inner"))
```
