```python
import numpy as np

# Word Vectors (Embeddings)
karak = np.array([0.9, 0.8, 0.2])
gahwa = np.array([0.8, 0.9, 0.1])
laptop = np.array([0.1, 0.0, 0.9])


def calculate_similarity(vec_a, vec_b):
    # Step 1: Calculate the dot product of vec_a and vec_b
    dot_prod = np.dot(vec_a, vec_b)

    # Step 2: Calculate the magnitudes (L2 Norm)
    mag_a = np.linalg.norm(vec_a)
    mag_b = np.linalg.norm(vec_b)

    # Step 3: Return the cosine similarity
    return dot_prod / (mag_a * mag_b)


print("Karak vs Gahwa:", calculate_similarity(karak, gahwa))
print("Karak vs Laptop:", calculate_similarity(karak, laptop))
# we create a function that take two vectors and get their dot product and their magnitudes and return the cosine similarity between them. We then use this function to compare the similarity between "Karak" and "Gahwa", and "Karak" and "Laptop".
```

```python
# Neural Network Weights (2x3 Matrix)
W = np.array([[1, -1, 2],
              [0, 3, 1]])

# Input features (3x1 Vector)
x = np.array([2,
              1,
              4])

# Step 1: Use np.dot() to multiply the Weight matrix by the input vector
output_y = np.dot(W, x)

print("Neural Network Output:", output_y)
# we use dot product to get the output of a simple neural network layer by multiplying the weight matrix W with the input vector x. The resulting output_y is a 2x1 vector that represents the activations of the neurons in the layer.
```

```python
# Encryption Key Matrix K
K = np.array([[4, 3],
              [3, 2]])

# Encrypted Message Vector C
C = np.array([[10],
              [7]])

# Step 1: Find the inverse of Matrix K
# Hint: Use np.linalg.inv()
K_inv = np.linalg.inv(K)

# Step 2: Multiply K_inv by C to get the Message M
message_M = np.dot(K_inv, C)

print("Decoded Message:\n", np.round(message_M))
# we have an encryption key matrix K and an encrypted message vector C. To decode the message, we first find the inverse of the matrix K using np.linalg.inv(), and then we multiply the inverse of K with the encrypted message C to retrieve the original message M. The result is rounded for better readability.
```
